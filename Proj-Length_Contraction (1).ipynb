### Determining Length Contractions
$$ \ell =\: Length\:  of\:  Rocket\:  at\:  rest $$
$$ d =\: Distance\: observed\: from\: Earth $$

When in motion, the Length of the Rocket as observed from Earth is:
$$ \ell_{e} = \ell\sqrt{1-\frac{v^2}{c^2}} $$

When in motion, the distance traveled as observed by the Pilot is:
$$ d_{p} = d\sqrt{1-\frac{v^2}{c^2}} $$

```python
vel_fix = Velocity / 100
Distance_PilotObs = Distance_EarthObs * math.sqrt(1-(vel_fix**2))
LengthRocket_EarthObs = LengthRocket_AtRest * math.sqrt(1-(vel_fix**2))

print(f"-The distance observed from Earth is the initial distance of {Distance_EarthObs} Light-Years.\n")
print(f"-While traveling, the distance observed by the Pilot is {Distance_PilotObs:.2f} Light-Years.\n")
print(f"    This is {((1 - (Distance_PilotObs/Distance_EarthObs)) * 100):.2f}% shorter than the distance observed at rest from Earth.\n")
print(f"-The Length of the Rocket in motion as observed from Earth is contracted to {LengthRocket_EarthObs:.2f} meters.\n")
print(f"    This is  {((1 - (LengthRocket_EarthObs/LengthRocket_AtRest)) * 100):.2f}% shorter than when it is observed at rest.")
```

## Determining Time Dilation



$$ Δt_{ob} = \frac{\ell_{E}}{v} $$


$$ Δt_{p} = \frac{\ell_{p}}{v} $$

```python
Time_EarthObs = Distance_EarthObs / Velocity
Time_PilotObs = Distance_PilotObs / Velocity

print(f"-Time experienced by the Earth-based observer is {Time_EarthObs:.2f} years.\n")
print(f"-Time experienced by the Pilot is {Time_PilotObs:.2f} years--{((1 - (Time_PilotObs/Time_EarthObs)) * 100):.2f}% shorter than the time experienced from Earth.")

#Graphic:
#Bar graphs comparing time 
```
